```python
#Linear regression class from scratch using OLS formula
class LR :
  def __init__ (self):
    self.m = 0
    self.b = 0

  def fit(self, X_train,y_train):
    #Calculating num and den of m
    num = 0
    den = 0
    for i in range(len(X_train)):  #Iterates over all i/p rows
      num += (X_train[i] - X_train.mean()) * (y_train[i] - y_train.mean())
      den += (X_train[i] - X_train.mean())**2
    self.m = num/ den
    self.b = y_train.mean()- (self.m *X_train.mean())
  def predict(self,x_test):
    y_pred = self.m*x_test+ self.b
    return y_pred

  def compute_cost(self,x_test,y_test):
    cost = 0
    for i in range(len(x_test)):
      cost += (y_test[i]- (self.m*x_test[i]) - self.b)**2
    return cost/ len(x_test)
```
