There are two main strategies:
* Replace with most frequent value -> mode.
* Replace with value like "missing" / or you can replace with random value.


## **Case-01 : Replace with mode**

* Data must be MCAR.
* Easy to implement
* Cons : Change the data distribution.

```python
df['col_name'].fillna(df['col_name'].mode()[0])
```

```python
df['col_name'].fillna('missing')
```
