# **Outlier detection using z-score:**

***Assumption*** : column must be normally distriuted or close to.

Within range of :
  1st std dev : 68%
  2st std dev : 95%
  3st std dev : 99%

Out of this range, lies outlier.

Z-score : xi`  = xi - u / std dev

**Treatment**:
* trimming
* capping

## **Approach-03 : Capping**

```python
upper_limit = df['col'].mean() + 3*df['col'].std()
lower_limit = df['col'].mean() - 3*df['col'].std()
```

```python
#main logic
df['col'] = np.where(
    df['col'] > upper_limit,
    np.where(
        df['col'] < lower_limit,
        lower_limit,
        df['col']
    )
)
```
