## **IQR Proximity Rule**

min = Q1 - 1.5IQR
max = Q3 + 1.5IQR
* used for skewed data

```python
df['col'].skew()
```

```python
#to check skewness
sns.boxplot(df['col'])
```

```python
#Finding IQR
per25 = df['col'].quantile(0.25)
per75 = df['col'].quantile(0.75) #or check using df.describe
```

```python
#Finding IQR
iqr = per75 - per25
```

```python
#Finding limits:
upper_limit = per75 + 1.5 * iqr
lower_limit = per25 - 1.5 * iqr
```

## Finding Outliers:

```python
df[df['col'] > upper_limit]
```

```python
df[df['col'] < lower_limit]
```

### Trimming:

```python
new_df = df [df['col'] < upper_limit]  #or
new_df = df [df['col'] > lower_limit]
```

### Capping:

```python
new_df_cap = df.copy()

new_df_cap['col'] = np.where(
    new_df_cap['col'] > upper_limit, #if true
    upper_limit, #replace with upper limit
    np.where( #else
        new_df_cap['col'] < lower_limit,
        lower_limit, #replace with lower limit
        new_df_cap['col'] #else leave it as it is
    )
)
```
