## **Code**:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('placement.csv')
```

```python
df.head()
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.xlabel("CGPA")
plt.ylabel("Salary Package")
```

```python
plt.boxplot(df)
```

```python
plt.boxplot(df['package'])
```

```python
plt.boxplot(df['cgpa'])
```

```python
x = df.iloc[:,0]
y = df.iloc[:,1]
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x.values.reshape(-1, 1),y,test_size=0.2,random_state=2) # Reshape x to a 2D array using reshape(-1, 1) during train_test_split
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
x_test
```

```python
lr.predict(x_test[0].reshape(1,1)) #checking for first cgpa
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.plot(x_test,lr.predict(x_test),color = 'red')
plt.xlabel("CGPA")
plt.ylabel("Salary Package")
```

* So our linear regression actually discovered or created this line.
* Mathematically for a line to be created, there must be a y-intercept and a slope.

```python
m = lr.coef_
q = lr.intercept_
m , q
```

y = mx + c , where x = cgpa

```python
#Proof
m* x_test[0] + q #the same value as predicted above
```

## **Intitution**
* For line it is actually finding m, q
* Where m represents the weightage of cgpa(how much the salary depend on cgpa), while q is just like some standard.
* Let's understand it with a scenario, where q = 0 => that means y = mx only i.e the higher the cgpa, the more the salary, if no cgpa no salary !!!but we have exceptions may be a person working w/o a degree => he ofcourse would be getting a salary here q plays it's role.
* Another case: experience (i/p), salary(o/p) so for fresher with experience = 0, no salary but they get a salary, a standard one = q.
